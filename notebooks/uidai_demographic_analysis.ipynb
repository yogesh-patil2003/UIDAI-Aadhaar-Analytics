```python
import pandas as pd
import os

folder = "/content/drive/MyDrive/uidai/api_data_aadhar_demographic/api_data_aadhar_demographic"

csv_files = [f for f in os.listdir(folder) if f.endswith('.csv')]

df_list = []
for f in csv_files:
    file_path = os.path.join(folder, f)
    temp = pd.read_csv(file_path)
    print(f"Loaded {f} – shape: {temp.shape}")
    df_list.append(temp)

# Combine into one DataFrame
df = pd.concat(df_list, ignore_index=True)
print("Combined shape:", df.shape)

df.head()
```

```python
df.info()
df.describe()
```

```python
df['date'] = pd.to_datetime(df['date'], dayfirst=True)
df['total_demo'] = df['demo_age_5_17'] + df['demo_age_17_']
```

```python
import matplotlib.pyplot as plt

daily = df.groupby('date')['total_demo'].sum().reset_index()

plt.figure(figsize=(10,5))
plt.plot(daily['date'], daily['total_demo'], marker='o')
plt.title("Total Demographic Count Over Time")
plt.xlabel("Date")
plt.ylabel("Total Count")
plt.grid(True)
plt.show()
```

```python
state_summary = df.groupby('state')[['demo_age_5_17', 'demo_age_17_']].sum()

state_summary.plot(kind='bar', figsize=(12,6))
plt.title("Demographic Age Groups by State")
plt.xlabel("State")
plt.ylabel("Count")
plt.show()
```

```python
dist_summary = df.groupby('district')[['demo_age_5_17', 'demo_age_17_']].sum()
dist_summary['total'] = dist_summary.sum(axis=1)

top10 = dist_summary.sort_values('total', ascending=False).head(10)

top10[['demo_age_5_17', 'demo_age_17_']].plot(kind='barh', figsize=(10,6))
plt.title("Top 10 Districts by Demographic Count")
plt.xlabel("Count")
plt.ylabel("District")
plt.show()
```

```python
import seaborn as sns

df['month'] = df['date'].dt.to_period('M')
heat = df.groupby(['state','month'])['total_demo'].sum().unstack(fill_value=0)

plt.figure(figsize=(12,8))
sns.heatmap(heat, cmap="YlGnBu", annot=True, fmt=".0f")
plt.title("State vs Month Demographic Heatmap")
plt.show()
```

```python
# Ensure date is datetime
df['date'] = pd.to_datetime(df['date'], dayfirst=True)

fig = px.line(df, x="date", y="total_demo", color="state",
              title="Total Demographic Count by State over Time")

fig.update_layout(
    xaxis=dict(
        rangeslider=dict(visible=True),
        type="date"
    )
)

fig.show()
```

```python
import plotly.graph_objects as go

fig = go.Figure()

states = df['state'].unique()

for state in states:
    df_state = df[df['state'] == state]
    fig.add_trace(
        go.Scatter(
            x=df_state['date'],
            y=df_state['total_demo'],
            name=state,
            visible=(state == states[0])  # first state is visible by default
        )
    )

buttons = []
for i, state in enumerate(states):
    buttons.append(dict(
        method="update",
        label=state,
        args=[{"visible": [j == i for j in range(len(states))]},
              {"title": f"Demographic Data for {state}"}]
    ))

fig.update_layout(
    updatemenus=[dict(active=0, buttons=buttons)]
)

fig.show()
```

```python
fig = px.line(df, x='date', y='total_demo', color='state',
              title="State Demographics with Interactivity")

fig.update_layout(
    updatemenus=[
        dict(
            active=0,
            buttons=[
                dict(label="All States",
                     method="update",
                     args=[{"visible": [True]*len(df['state'].unique())},
                           {"title": "All States"}])
            ] + [
                dict(label=state,
                     method="update",
                     args=[{"visible": df['state'] == state},
                           {"title": f"{state} Demographic Trend"}])
                for state in df['state'].unique()
            ]
        )
    ],
    xaxis=dict(
        rangeslider=dict(visible=True),
        type="date"
    )
)

fig.show()
```
